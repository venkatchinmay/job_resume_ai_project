## 3. Convert Prompt to JSON

````python
import json
import re
from typing import Optional, Dict, Any
def extract_and_parse_json(
    text: str, 
    strict: bool = False
) -> Optional[Dict[Any, Any]]:
    """
    Extract and parse JSON from various text formats
    
    Args:
        text: Input text containing JSON
        strict: If True, raise exceptions; if False, return None on error
    
    Returns:
        Parsed JSON dictionary or None
    """
    try:
        # Method 1: Try markdown code block
        pattern = r'```(?:json)?\s*(\{.*?\})\s*```'
        match = re.search(pattern, text, re.DOTALL)
        
        if match:
            return json.loads(match.group(1))
        
        # Method 2: Try to find raw JSON
        start = text.find('{')
        end = text.rfind('}') + 1
        
        if start != -1 and end != 0:
            json_str = text[start:end]
            return json.loads(json_str)
        
        if strict:
            raise ValueError("No JSON found in text")
        return None
        
    except json.JSONDecodeError as e:
        if strict:
            raise ValueError(f"Invalid JSON: {e}")
        return None
    except Exception as e:
        if strict:
            raise ValueError(f"Error extracting JSON: {e}")
        return None
# Usage
data = extract_and_parse_json(response.content)
if data:
    print(json.dumps(data, indent=2))
else:
    print("Failed to extract JSON")
````

## 2. Define a Tool
We will define a custom tool that the LLM can use. For this example, let's create a tool to calculate years of experience derived from the text, or just a simple arithmetic tool.

````python
@tool
def calculate_experience_score(years: int) -> int:
    """Calculates a score based on years of experience."""
    return years * 10

tools = [calculate_experience_score]
````

## 3. Configure LLM with Tools
Now we fetch a free model from OpenRouter and bind the tools to it.

````python
# Get available free models
free_models = get_free_models()
if not free_models:
    print("No free models available. Check your API key or connectivity.")
else:
    model_name = free_models[0]
    print(f"Using model: {model_name}")

    # Configure LLM arguments
    platform = "openrouter"
    temperature = 0.0
    arguments = {
        "prompt": "You are a helpful assistant. Use the calculate_experience_score tool to score the candidate's experience."
    }

    # Get the model with tools bound
    chat_model, prompt = LLM.get_llm_model_with_tools(
        platform=platform,
        model_name=model_name,
        temperature=temperature,
        arguments=arguments,
        tools=tools
    )
    
    print(f"Model bound with: {chat_model.tools}")
````

## 4. Run the Chain
We send a user query that triggers the tool.

````python
from langchain_core.messages import HumanMessage, SystemMessage, ToolMessage
import json

# 1. Create a map to easily look up tools by name
tool_map = {t.name: t for t in tools}

if free_models:
    messages = [
        SystemMessage(content=arguments['prompt']),
        HumanMessage(content="The candidate has 5 years of experience. What is their score?")
    ]

    # 2. First Invoke: LLM decides if it needs a tool
    ai_msg = chat_model.invoke(messages)
    messages.append(ai_msg) # Add AI's initial response to history
    
    # 3. Check if LLM wants to use tools
    if ai_msg.tool_calls:
        print(f"LLM wants to use tools: {len(ai_msg.tool_calls)}")
        
        # 4. Iterate through all requested tool calls
        for tool_call in ai_msg.tool_calls:
            selected_tool = tool_map[tool_call["name"].lower()]
            
            # Execute the tool!
            print(f"Invoking tool: {tool_call['name']} with args: {tool_call['args']}")
            tool_output = selected_tool.invoke(tool_call["args"])
            
            # 5. Create a ToolMessage with the result and add to history
            messages.append(ToolMessage(
                content=str(tool_output), 
                tool_call_id=tool_call["id"]
            ))
            
        # 6. Second Invoke: Get the final answer from LLM with the tool results
        final_response = chat_model.invoke(messages)
        
        print("\nFinal Answer:", final_response.content)
        
    else:
        print("Final Answer:", ai_msg.content)
````
